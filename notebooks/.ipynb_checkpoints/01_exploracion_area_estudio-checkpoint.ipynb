```python
# Bibliotecas para análisis y visualización geográfica

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import cartopy.crs as ccrs
import cartopy.feature as cfeature

print("Bibliotecas importadas correctamente")
```

```python
# Definir los límites geográficos del área de estudio

latitud_min = 42
latitud_max = 46

longitud_min = 143
longitud_max = 148

print("Delimitación del área de estudio:")
print("Latitud:", latitud_min, "a", latitud_max, "°N")
print("Longitud:", longitud_min, "a", longitud_max, "°E")
```

```python
# Crear el mapa base del noreste de Hokkaido, Japón

fig = plt.figure(figsize=(12, 8))

ax = plt.axes(
    projection=ccrs.PlateCarree()
)

# Definir la extensión geográfica visible
ax.set_extent(
    [142.5, 148.5, 41.5, 46.5],
    crs=ccrs.PlateCarree()
)

# Agregar océano y superficie terrestre
ax.add_feature(cfeature.OCEAN, facecolor="lightblue")
ax.add_feature(cfeature.LAND, facecolor="lightgray")

# Agregar líneas de costa
ax.add_feature(cfeature.COASTLINE, linewidth=0.8)

# Agregar fronteras
ax.add_feature(cfeature.BORDERS, linestyle=":")

# Título
ax.set_title(
    "Región de estudio: noreste de Hokkaido, Japón",
    fontsize=14
)

# Mostrar mapa
plt.show()
```

```python
# Dibujar los límites del área de estudio

ax.plot(
    [longitud_min, longitud_max, longitud_max,
     longitud_min, longitud_min],
    [latitud_min, latitud_min, latitud_max,
     latitud_max, latitud_min],
    color="red",
    linewidth=2,
    linestyle="--",
    transform=ccrs.PlateCarree(),
    label="Área de estudio"
)

ax.legend(loc="upper left")

plt.show()
```

```python
# Crear mapa con el área de estudio delimitada

fig = plt.figure(figsize=(12, 8))

ax = plt.axes(projection=ccrs.PlateCarree())

# Extensión del mapa
ax.set_extent(
    [142.5, 148.5, 41.5, 46.5],
    crs=ccrs.PlateCarree()
)

# Elementos geográficos
ax.add_feature(cfeature.OCEAN, facecolor="lightblue")
ax.add_feature(cfeature.LAND, facecolor="lightgray")
ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linestyle=":")

# Delimitación del área de estudio
ax.plot(
    [143, 148, 148, 143, 143],
    [42, 42, 46, 46, 42],
    color="red",
    linewidth=2.5,
    linestyle="--",
    transform=ccrs.PlateCarree(),
    label="Área de estudio"
)

# Título y leyenda
ax.set_title(
    "Área de estudio ampliada: noreste de Hokkaido, Japón",
    fontsize=14
)

ax.legend(loc="upper left")

plt.show()
```
