```python
#Coincide el sistema de coordenadas, se visualiza la extensión del polígono de GEOJSON y de los datos de clorofila
# Seleccionar clorofila del primer mes
chl_enero = ds["CHL"].isel(time=0)

# Revisar la fecha
print(ds.time.values[0])
```

### Prueba para visualizar que los datos de clorofila y el polígono del AMP coincidan en zonas geográficas

```python
#Visualizar clorfoila y ambos polígonos 

# Crear figura
fig, ax = plt.subplots(figsize=(11, 8))

# 1. Mostrar raster de clorofila
chl_enero.plot(
    ax=ax,
    x="longitude",
    y="latitude",
    cmap="viridis",
    cbar_kwargs={"label": "Clorofila-a"}
)

# 2. Superponer los polígonos EBSA
poligono.boundary.plot(
    ax=ax,
    color="black",
    linewidth=2,
    label="Límites EBSA"
)

# 3. Agregar nombres de las áreas
for _, row in poligono.iterrows():
    centro = row.geometry.representative_point()

    ax.annotate(
        row["AreaName"],
        (centro.x, centro.y),
        fontsize=8,
        color="white",
        weight="bold",
        ha="center"
    )

# 4. Configurar mapa
ax.set_title("Clorofila-a y áreas EBSA de Shiretoko - Enero 2021")
ax.set_xlabel("Longitud (°E)")
ax.set_ylabel("Latitud (°N)")

plt.tight_layout()
plt.show()
```

```python



#Preparar los datos de clorofila, genera un DataArray de clorofila identificado como raster, con dimensiones espaciales y CRS definidos

da = ds["CHL"]

da = da.rio.set_spatial_dims(
    x_dim="longitude",
    y_dim="latitude"
)

da = da.rio.write_crs("EPSG:4326")
```

```python
print(da.dims)
print(da.rio.crs)
print(da.shape)
```

### Recorte de los datos de clorofila a polígono del AMP

```python
#Recorte de los datos
chl_clip = da.rio.clip(
    poligono.geometry,
    poligono.crs,
    drop=True
)
```

```python
print("Datos originales:")
print(da.shape)

print("Datos recortados:")
print(chl_clip.shape)
```

```python
# Contar píxeles válidos del primer mes
pixeles_validos = chl_clip.isel(time=0).count().values

# Contar el total de posiciones espaciales
pixeles_totales = (
    chl_clip.sizes["latitude"] *
    chl_clip.sizes["longitude"]
)

print("Píxeles totales:", pixeles_totales)
print("Píxeles válidos:", pixeles_validos)
print("Píxeles enmascarados:", pixeles_totales - pixeles_validos)
```

```python
#Distribución de pixeles válidos

fig, ax = plt.subplots(figsize=(9, 7))

chl_clip.isel(time=0).plot(
    ax=ax,
    cmap="viridis",
    cbar_kwargs={"label": "Clorofila-a"}
)

poligono.boundary.plot(
    ax=ax,
    color="red",
    linewidth=1.5
)

ax.set_title("Distribución de píxeles válidos de CHL - Enero 2021")
plt.show()
```

```python
# Contar píxeles válidos para cada uno de los 60 meses
pixeles_por_mes = chl_clip.count(
    dim=["latitude", "longitude"]
)

# Convertir a DataFrame
pixeles_df = pixeles_por_mes.to_dataframe(
    name="pixeles_validos"
).reset_index()

print(pixeles_df.head(20))
```

```python
# Identificar valores únicos de píxeles válidos
print(pixeles_df["pixeles_validos"].value_counts().sort_index())
```

```python
# Identificar meses con menos de 33 píxeles válidos en todo el set de datos

meses_incompletos = pixeles_df[
    pixeles_df["pixeles_validos"] < 33
]

print(meses_incompletos)
```

### Inicio de análisis de datos del polígono:  Copernicus ya proporciona composiciones mensuales de clorofila, ahora se calculará el promedio espacial de los pixeles del nuevo rectángulo para cada mes

```python


# Convertir NetCDF recortado a DataFrame

chl_df = chl_rect.to_dataframe(
    name="CHL"
).reset_index()

# Convertir fecha a formato datetime
chl_df["time"] = pd.to_datetime(chl_df["time"])

# Crear columnas de año y mes
chl_df["year"] = chl_df["time"].dt.year
chl_df["month"] = chl_df["time"].dt.month

# Revisar información
print(chl_df.head())
print("Total de registros:", len(chl_df))
```

```python
# Promedio espacial por mes y año
chl_mensual = (
    chl_df
    .groupby(["year", "month"])["CHL"]
    .mean()
    .reset_index()
)
```

```python
print(chl_mensual.head(12))
```

```python
#Comprobación de disponibilidad de datos para los 60 meses 
print("Total de registros:", len(chl_mensual))

print("Años disponibles:", chl_mensual["year"].unique())

print("Meses disponibles:", chl_mensual["month"].nunique())
```

```python
#Organización de datos en tabla comparativa
tabla_chl = chl_mensual.pivot(
    index="month",
    columns="year",
    values="CHL"
)

print(tabla_chl)
```

```python
# Exportar tabla a CSV
tabla_chl.to_csv(
    "promedio_mensual_clorofila_Shiretoko_2021_2025.csv",
    index=True,
    index_label="month"
)
```

### Unificación de análisis de clorofila con registros de esfuerzo pesquero aparente y horas de esfuerzo pesquero aparente por año

```python

#Se importa el csv de los datos ya procesados de GFW y se revisa su estructura
import pandas as pd

gfw_df = pd.read_csv("GFW_Shiretoko_2020_2024.csv")

print(gfw_df.head())
print(gfw_df.columns)
print(gfw_df.shape)
```

```python
# Convertir la columna date a formato fecha
gfw_df["date"] = pd.to_datetime(
    gfw_df["date"],
    format="%Y-%m"
)

# Identificar los meses disponibles por año
meses_disponibles = (
    gfw_df.groupby(gfw_df["date"].dt.year)["date"]
    .apply(lambda x: sorted(x.dt.month.unique()))
)

display(meses_disponibles)
```

### Considerando los resultados del heat map, se propone hacer la comparación entre el mes de abril y el esfuerzo pesquero porque es en el que se presenta una mayor concentración promedio de clorofila.Sin embargo, no hay datos disponibles para 2022

```python


# Filtrar abril

gfw_abril = gfw_df[
    gfw_df["date"].dt.month == 4
].copy()

# Agrupar por año y calcular los dos indicadores
resumen_abril = (
    gfw_abril
    .groupby(gfw_abril["date"].dt.year)
    .agg(
        registros_pesca=("hours", "count"),
        intensidad_pesca_horas=("hours", "sum")
    )
)

# Incluir todos los años del periodo de estudio
# NaN indica que no se encontraron registros en ese año
resumen_abril = resumen_abril.reindex(
    [2021, 2022, 2023, 2024]
)

# Mostrar resultados
display(resumen_abril.round(2))
```

```python
# Seleccionar únicamente los años de interés (por falta de datos)

gfw_abril_comparacion = gfw_abril[
    gfw_abril["date"].dt.year.isin([2021, 2023, 2024])
].copy()


# Agrupar por año y ubicación geográfica
gfw_mapa = (
    gfw_abril_comparacion
    .groupby(
        [
            gfw_abril_comparacion["date"].dt.year.rename("year"),
            "lat",
            "lon"
        ],
        as_index=False
    )
    .agg(
        registros_pesca=("hours", "count"),
        intensidad_pesca_horas=("hours", "sum")
    )
)

# Mostrar resultados
display(gfw_mapa)
```

```python
# Seleccionar abril de 2021, 2023 y 2024
gfw_abril_comparacion = gfw_abril[
    gfw_abril["date"].dt.year.isin([2021, 2023, 2024])
].copy()

# Agrupar por año y ubicación
gfw_mapa = (
    gfw_abril_comparacion
    .groupby(
        [
            gfw_abril_comparacion["date"].dt.year.rename("year"),
            "lat",
            "lon"
        ],
        as_index=False
    )
    .agg(
            registros_pesca=("hours", "count"),
    intensidad_pesca_horas=("hours", "sum")
    )
)

display(gfw_mapa)
```

```python
#Preparación de mapa de intesidad y esfuerzo aparente

import matplotlib.pyplot as plt

# Seleccionar abril de 2021
gfw_2021 = gfw_mapa[gfw_mapa["year"] == 2021]

# Crear mapa
fig, ax = plt.subplots(figsize=(9, 7))

# Representar registros e intensidad
sc = ax.scatter(
    gfw_2021["lon"],
    gfw_2021["lat"],
    c=gfw_2021["intensidad_pesca_horas"],
    s=100 + gfw_2021["registros_pesca"] * 100,
    cmap="YlOrRd",
    edgecolor="black",
    alpha=0.9
)

# Barra de color: intensidad
cbar = plt.colorbar(sc, ax=ax)
cbar.set_label("Intensidad de pesca aparente (horas)")

#Leyenda para indicar que el tamaño de los círculos corresponde al número de registros

valores_registros = sorted(
    gfw_2021["registros_pesca"].unique()
)

handles = []

for n in valores_registros:
    handles.append(
        ax.scatter(
            [],
            [],
            s=100 + n * 100,
            facecolors="none",
            edgecolors="black",
            label=f"{n} registro(s)"
        )
    )

ax.legend(
    handles=handles,
    title="Registros de pesca aparente",
    loc="upper right",
    frameon=True
)

#Etiquetas y formato

ax.set_xlabel("Longitud (°E)")
ax.set_ylabel("Latitud (°N)")

ax.set_title(
    "Pesca aparente - Abril 2021",
    fontsize=13,
    fontweight="bold"
)

ax.set_xlim(144.5, 145.8)
ax.set_ylim(43.7, 44.8)

ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Integración de datos de clorofila con parámetros de pesca en abril de 2021, 2023 y 2024, individualmente y en conjunto para su comparación

```python
# Integración de clorofila con parámetros de pesca para abril 2021

import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

# Selección de datos

gfw_2021 = gfw_mapa[
    gfw_mapa["year"] == 2021
].copy()

chl_abril_2021 = chl_rect.sel(
    time="2021-04"
).squeeze()

# Configurar escalas comunes

# Escala de clorofila
vmin_chl = 1.81
vmax_chl = 7.13

# Escala de registros de pesca aparente
max_registros = max(
    2,
    int(gfw_mapa["registros_pesca"].max())
)

norm_registros = Normalize(
    vmin=1,
    vmax=max_registros
)

cmap_registros = plt.cm.YlOrRd

# Función para escalar el tamaño de los círculos según las horas
def escalar_horas(horas):
    return 70 + horas * 35

# Generación de figura

fig, ax = plt.subplots(figsize=(13, 8))

fig.subplots_adjust(
    left=0.08,
    right=0.72,
    top=0.88,
    bottom=0.12
)

# Raster de clorofila

im = ax.pcolormesh(
    chl_abril_2021["longitude"],
    chl_abril_2021["latitude"],
    chl_abril_2021.values,
    cmap="viridis",
    vmin=vmin_chl,
    vmax=vmax_chl,
    shading="auto",
    rasterized=True
)

# Incorporación de número de registros e intensidad de pesca

# Tamaño según horas acumuladas
tamanos = escalar_horas(
    gfw_2021["intensidad_pesca_horas"]
)

# Color según número de registros
colores = gfw_2021["registros_pesca"]

sc = ax.scatter(
    gfw_2021["lon"],
    gfw_2021["lat"],
    s=tamanos,
    c=colores,
    cmap=cmap_registros,
    norm=norm_registros,
    edgecolor="black",
    linewidth=0.8,
    alpha=0.95,
    zorder=3
)

# Barra de color para leyenda de clorofila

cax_chl = fig.add_axes([0.76, 0.25, 0.025, 0.50])

cbar_chl = fig.colorbar(im, cax=cax_chl)

cbar_chl.set_label(
    "Clorofila-a (mg m⁻³)",
    fontsize=9
)

# Barra de color para número de registros de pesca aparente

cax_pesca = fig.add_axes([0.83, 0.25, 0.025, 0.50])

cbar_pesca = fig.colorbar(sc, cax=cax_pesca)

cbar_pesca.set_label(
    "Registros de pesca aparente",
    fontsize=9
)

cbar_pesca.set_ticks(
    range(1, max_registros + 1)
)

# Leyenda de tamaño de los círculos según las horas acumuladas

valores_horas = [1, 4, 7]

handles = []

for horas in valores_horas:

    handle = ax.scatter(
        [],
        [],
        s=escalar_horas(horas),
        facecolors="none",
        edgecolors="black",
        linewidth=0.8,
        label=f"{horas} horas"
    )

    handles.append(handle)

fig.legend(
    handles=handles,
    title="Intensidad de pesca aparente",
    loc="upper left",
    bbox_to_anchor=(0.87, 0.85),
    frameon=False,
    fontsize=9,
    title_fontsize=9
)

# Delimitación de extensión geográfica

ax.set_xlim(144.5, 145.8)
ax.set_ylim(43.7, 44.8)

ax.set_xlabel("Longitud (°E)", fontsize=10)
ax.set_ylabel("Latitud (°N)", fontsize=10)

# Título y formato

ax.set_title(
    "Clorofila-a y pesca aparente | Abril 2021",
    fontsize=13,
    fontweight="bold",
    pad=14
)

ax.tick_params(labelsize=9)
ax.grid(alpha=0.12, linestyle="--")

plt.show()
```

### Análisis de la serie temporal de clorofila y esfuerzo pesquero 

```python
#Pasemos al análisis de la serie temporal de clorofila y esfuerzo pesquero

# Preparar datos de esfuerzo pesquero

gfw_df["date"] = pd.to_datetime(gfw_df["date"])

gfw_df["year"] = gfw_df["date"].dt.year
gfw_df["month"] = gfw_df["date"].dt.month

# Filtrar periodo de coincidencia: 2021-2024
gfw_periodo = gfw_df[
    gfw_df["year"].between(2021, 2024)
].copy()

# Agregar horas por mes
gfw_mensual = (
    gfw_periodo
    .groupby(["year", "month"], as_index=False)
    .agg(
        horas_totales=("hours", "sum"),
        registros=("hours", "count")
    )
)

display(gfw_mensual)
```

```python
#Integración de clorofila y esfuerzo pesquero para Shiretoko, Japón (2021-2024)

# Filtrar clorofila para el periodo de coincidencia
chl_periodo = chl_mensual[
    chl_mensual["year"].between(2021, 2024)
].copy()

# Integrar las dos bases mediante año y mes
datos_integrados = pd.merge(
    chl_periodo,
    gfw_mensual,
    on=["year", "month"],
    how="left"
)

# Los meses sin registros se representan con cero
datos_integrados["horas_totales"] = (
    datos_integrados["horas_totales"].fillna(0)
)

datos_integrados["registros"] = (
    datos_integrados["registros"].fillna(0)
)

# Crear columna de fecha
datos_integrados["fecha"] = pd.to_datetime(
    dict(
        year=datos_integrados["year"],
        month=datos_integrados["month"],
        day=1
    )
)

# Ordenar cronológicamente
datos_integrados = datos_integrados.sort_values(
    "fecha"
).reset_index(drop=True)

display(datos_integrados)
```
